```python
import json

objects = json.load(open("../objects/objects.json", "r"))
```

```python
import starplot
import starplot.models
import starplot.styles
import math

style = starplot.styles.PlotStyle().extend(
    #starplot.styles.extensions.BLUE_LIGHT,
    starplot.styles.extensions.MAP,
    font_size=36,
)

for objectstyle in [
    style.dso_association_stars,
    style.dso_dark_nebula,
    style.dso_double_star,
    style.dso_duplicate,
    style.dso_galaxy,
    style.dso_globular_cluster,
    style.dso_nebula,
    style.dso_nonexistant,
    style.dso_nova_star,
    style.dso_open_cluster,
    style.dso_planetary_nebula,
    style.dso_supernova_remnant,
]:
    objectstyle.marker.fill = "rgba(0, 0, 0, 0.07)"
    objectstyle.marker.stroke_width = 10

def makechart(prefix, thisname, object, fov, diameterstep=1):

    print(thisname)

    p = starplot.MapPlot(
        projection=starplot.Stereographic(
            center_ra=object["alpha"],
            center_dec=object["delta"],
        ),
        ra_min=object["alpha"] - fov / 2 / math.cos(math.radians(object["delta"])),
        ra_max=object["alpha"] + fov / 2 / math.cos(math.radians(object["delta"])),
        dec_min=object["delta"] - fov / 2,
        dec_max=object["delta"] + fov / 2,
        style=style,
        autoscale=True,
        resolution=1200,
    )


    p.stars(
        where=[starplot._.magnitude < 9],
        where_labels=False,
        opacity_fn=lambda star: 1.0,
        size_fn=lambda star: (10 - star.magnitude) * 12,
    )

    p.dsos(
        where_labels=False,
    )

    circlestyle = starplot.styles.PolygonStyle()
    circlestyle.fill = None
    circlestyle.stroke_width=2
    circlestyle.dash_array=(30,30,30)
    circlestyle.opacity=0.3
    for i in range(diameterstep, int(fov) + 1, diameterstep):
        p.circle(
            center=(object["alpha"],object["delta"]),
            radius_degrees=i/2,
            style=circlestyle
            )

    p.export("%s/%s-chart.svg" % (prefix, thisname))
    del p

def makecharts(prefix, thiscatalog, ifirst, ilast, fov, diameterstep=1):
    for i in range(ifirst, ilast + 1):
        for id in objects:
            object = objects[id]
            if "Catalog %s" % thiscatalog not in object:
                continue
            if object["Catalog %s" % thiscatalog] != i:
                continue

            thisname = "%s%d" % (thiscatalog, object["Catalog %s" % thiscatalog])

            makechart(prefix, thisname, object, fov, diameterstep=diameterstep)

if False:

    !mkdir -p ../binocular/figures
    makecharts("../binocular/figures", "BSC", 16, 16, 6, diameterstep=2)
    !sh svgtopdf.sh ../binocular/figures/BSC16-chart.svg

    !mkdir -p ../finder/figures
    makecharts("../finder/figures", "M", 16, 16, 6, diameterstep=2)
    !sh svgtopdf.sh ../finder/figures/M16-chart.svg
```
