## Process objects-raw.json to give objects.json

```python
import json

objectdict = json.load(open("objects-raw.json", "r"))

for id in objectdict.keys():
    valuesdictarray = objectdict[id]
    valuesdict = dict()
    for i in range(len(valuesdictarray)):
        valuesdict |= valuesdictarray[i]
    provenances = set()
    for i in range(len(valuesdictarray)):
        provenances.add(valuesdictarray[i]["Provenance"])
    valuesdict["Provenances"] = list(provenances)
    objectdict[id] = valuesdict

json.dump(objectdict, open("objects.json", "w"), indent=2)
```

## Make summary tables

```python
import json
import sys
import contextlib

objects = json.load(open("objects.json", "r"))

tableheader = r"""
\begin{table}[t]
    \setlength{\tabcolsep}{2pt}
    \footnotesize
    \begin{tabularx}{\linewidth}{lccX}
        \hline
        Name&Position&Type&Other Names\\
        \hline
"""

tablefooter = r"""
        \hline
    \end{tabularx}
\end{table}
\clearpage
"""

nlines = 50

def maketable(prefix, thiscatalog, ifirst, ilast, exceptions=[]):
    with open("%s/%s-table.tex" % (prefix, thiscatalog), "w") as f:
        iline = 0
        with contextlib.redirect_stdout(f):
            for i in range(ifirst, ilast + 1):
                if i in exceptions:
                    continue
                if iline % nlines == 0:
                    print(tableheader)
                for id in objects:
                    object = objects[id]
                    if "Catalog %s" % thiscatalog not in object:
                        continue
                    if object["Catalog %s" % thiscatalog] != i:
                        continue
                    if thiscatalog in ["M", "C", "U"]:
                        thisname = "%s%d" % (thiscatalog, object["Catalog %s" % thiscatalog])
                    else:
                        thisname = "%s %d" % (thiscatalog, object["Catalog %s" % thiscatalog])
                    othercatalogs = ["M", "C", "U", "BSC", "NGC", "IC", "Mel", "Cr"]
                    othernames = []
                    for othercatalog in othercatalogs:
                        if (
                            thiscatalog != othercatalog
                            and "Catalog %s" % othercatalog in object
                        ):
                            if othercatalog in ["M", "C", "U"]:
                                othernames.append(
                                    "%s%s"
                                    % (othercatalog, object["Catalog %s" % othercatalog])
                                )
                            else:
                                othernames.append(
                                    "%s %s"
                                    % (othercatalog, object["Catalog %s" % othercatalog])
                                )
                    if "Name" in object:
                        othernames.append(object["Name"])
                    othernames = " = ".join(othernames)
                    line = "%-4s &$%04.1f$ $%+03.0f$&%-3s&%s\\\\" % (
                        thisname,
                        object["alpha"] / 15,
                        object["delta"],
                        object["Type"],
                        othernames,
                    )
                    print(line, file=sys.stderr)
                    print(line)
                    iline += 1
                    if iline % nlines == 0:
                        print(tablefooter)
            if iline % nlines != 0:
                print(tablefooter)

maketable("../finder/", "M", 1, 110)
maketable("../finder/", "C", 1, 109)
maketable("../finder/", "U", 1, 87)
maketable("../binocular/", "BSC", 1, 150, [2, 8, 71])
```

## Make Bodies

```python
import json
import sys
import contextlib

objects = json.load(open("objects.json", "r"))

tableheader = r"""
\begin{table}[t]
    \setlength{\tabcolsep}{2pt}
    \footnotesize
    \begin{tabularx}{\linewidth}{lccX}
        \hline
        Name&Position&Type&Other Names\\
        \hline
"""

tablefooter = r"""
        \hline
    \end{tabularx}
\end{table}
\clearpage
"""


def makebody(prefix, thiscatalog, ifirst, ilast):
    with open("%s/%s-body.tex" % (prefix, thiscatalog), "w") as f:
        iline = 0
        with contextlib.redirect_stdout(f):
            for i in range(ifirst, ilast + 1):
                for id in objects:

                    object = objects[id]

                    if "Catalog %s" % thiscatalog not in object:
                        continue
                    if object["Catalog %s" % thiscatalog] != i:
                        continue

                    if thiscatalog in ["M", "C", "U"]:
                        thisname = "%s%d" % (
                            thiscatalog,
                            object["Catalog %s" % thiscatalog],
                        )
                    else:
                        thisname = "%s %d" % (
                            thiscatalog,
                            object["Catalog %s" % thiscatalog],
                        )
                    thischart = "%s%d" % (
                        thiscatalog,
                        object["Catalog %s" % thiscatalog],
                    )

                    othercatalogs = ["M", "C", "U", "BSC", "NGC", "IC", "Mel", "Cr"]
                    othernames = []
                    for othercatalog in othercatalogs:
                        if (
                            thiscatalog != othercatalog
                            and "Catalog %s" % othercatalog in object
                        ):
                            if othercatalog in ["M", "C", "U"]:
                                othernames.append(
                                    "%s%s"
                                    % (
                                        othercatalog,
                                        object["Catalog %s" % othercatalog],
                                    )
                                )
                            else:
                                othernames.append(
                                    "%s %s"
                                    % (
                                        othercatalog,
                                        object["Catalog %s" % othercatalog],
                                    )
                                )
                    if "Name" in object:
                        othernames.append(object["Name"])
                    othernames = " = ".join(othernames)

                    X = object["X"]
                    if "Y" not in object:
                        Y = 0
                    else:
                        Y = object["Y"]
                    if X >= 180:
                        if X == Y or Y == 0:
                            size = r"\degrees{%.1f}" % (X / 60)
                        else:
                            size = r"\degrees{%.1f} \times \degrees{%.1f}" % (
                                X / 60,
                                Y / 60,
                            )
                    else:
                        if X == Y or Y == 0:
                            size = r"\arcmin{%.1f}" % (X)
                        else:
                            size = r"\arcmin{%.1f} \times \arcmin{%.1f}" % (X, Y)

                    psa = object["PSA"]
                    if "OSA" in object:
                        osa = object["OSA"]
                    else:
                        osa = ""
                    if "CSA" in object:
                        csa = object["CSA"]
                    else:
                        csa = ""
                    if "Constellation" in object:
                        constellation = object["Constellation"]
                    else:
                        constellation = ""
                    if "Mag" in object:
                        mag = object["Mag"]
                    else:
                        mag = ""
                    if "Hops" in object:
                        hops = object["Hops"]
                    else:
                        hops = ""
                    if "Notes" in object:
                        notes = "Notes: %s" % object["Notes"]
                    else:
                        notes = ""

                    references = []
                    for provenance in [
                        "OSA",
                        "DSC:TMO",
                        "DSC:TCO",
                        "DSC:HT",
                        "DSC:TSD",
                        "DSC:SG",
                    ]:
                        if provenance in object["Provenances"]:
                            references.append(provenance)
                    references = " ".join(references)

                    line = (
                        r"\chart{%s}{%s}{%s}{\chartdata{%04.1f}{%+03.0f}{%s}{%s}{%s}{%s}{%s}{%s}{%s}}{%s}{%s}{%s}"
                        % (
                            thisname,
                            thischart,
                            othernames,
                            object["alpha"] / 15,
                            object["delta"],
                            constellation,
                            psa,
                            osa,
                            csa,
                            object["Type"],
                            mag,
                            size,
                            references,
                            hops,
                            notes,
                        )
                    )
                    print(line, file=sys.stderr)
                    print(line)


makebody("../finder", "M", 1, 110)
makebody("../finder", "C", 1, 109)
makebody("../finder", "U", 1, 87)
makebody("../binocular", "BSC", 1, 150)
```
